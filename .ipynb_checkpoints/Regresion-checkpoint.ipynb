```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

# Configurar estilo de gráficos
plt.style.use('default')

# =============================================================================
# 1. CARGA Y PREPARACIÓN DE DATOS ESENCIAL
# =============================================================================
print("=" * 80)
print("PREPARACIÓN DE DATOS PARA REGRESIÓN LINEAL")
print("=" * 80)

# Cargar el dataset (Asegúrate de que 'world_bank_data_2025.csv' esté en la ruta)
try:
    df = pd.read_csv('world_bank_data_2025.csv')
except FileNotFoundError:
    print("❌ ERROR: Asegúrate de que el archivo 'world_bank_data_2025.csv' esté en la misma carpeta.")
    # Crear un DataFrame simulado para que el código no falle si el archivo no existe
    n = 3472
    np.random.seed(42)
    df = pd.DataFrame({
        'country_name': np.random.choice(['CountryA', 'CountryB', 'CountryC'], n),
        'Unemployment Rate (%)': np.random.normal(7.44, 5.41, n),
        'GDP Growth (% Annual)': np.random.normal(2.89, 5.54, n),
        'Inflation (CPI %)': np.random.normal(5.56, 17.42, n),
        'GDP (Current USD)': np.exp(np.random.normal(25, 3, n)) # Simulación de PIB log-normal
    })
    # Introducir algunos nulos para simular la limpieza
    df.loc[df.sample(frac=0.05).index, 'Inflation (CPI %)'] = np.nan
    df.loc[df.sample(frac=0.03).index, 'Unemployment Rate (%)'] = np.nan
    print("\n⚠️ Usando DataFrame simulado. Reemplaza con tu carga de datos real.")

# Definir variables clave
economic_cols = [
    'Unemployment Rate (%)', 'GDP Growth (% Annual)', 
    'Inflation (CPI %)', 'GDP (Current USD)'
]

# 2.2. Imputación de valores faltantes (Mediana por país)
print("\n✅ IMPUTACIÓN DE VALORES FALTANTES (Mediana por País):")
for col in economic_cols:
    if col in df.columns and df[col].isnull().sum() > 0:
        initial_nulls = df[col].isnull().sum()
        df[col] = df.groupby('country_name')[col].transform(
            lambda x: x.fillna(x.median() if not np.isnan(x.median()) else df[col].median())
        )
        final_nulls = df[col].isnull().sum()
        print(f"   - {col}: {initial_nulls} → {final_nulls} nulos")

# 2.3. Eliminar filas con nulos persistentes (si los hay tras la imputación)
df.dropna(subset=economic_cols, inplace=True)
print(f"\n✅ Filas totales después de limpieza: {len(df)}")


# =============================================================================
# 2. TRANSFORMACIÓN Y SELECCIÓN DE VARIABLES PARA REGRESIÓN
# =============================================================================
print("\n\n2. TRANSFORMACIÓN Y SELECCIÓN DE VARIABLES")
print("-" * 40)

# 2.1. Transformación Logarítmica para el PIB
# Se necesita el logaritmo para manejar la escala y alta varianza (CV 476.05%)
# (Como se hizo en tu análisis bootstrap)
# Filtrar solo valores positivos antes del log
df['PIB_log'] = np.log(df['GDP (Current USD)'][df['GDP (Current USD)'] > 0])
df.dropna(subset=['PIB_log'], inplace=True) # Eliminar registros donde no se pudo aplicar log
print("✅ Variable 'GDP (Current USD)' transformada a 'PIB_log'.")

# 2.2. Definición del modelo (Regresión Lineal Múltiple)
# Y: Tasa de Desempleo (Variable dependiente)
Y = df['Unemployment Rate (%)']

# X: Crecimiento, Inflación y PIB_log (Variables independientes)
X_vars = ['GDP Growth (% Annual)', 'Inflation (CPI %)', 'PIB_log']
X = df[X_vars]

print(f"✅ Variable Dependiente (Y): {Y.name}")
print(f"✅ Variables Independientes (X): {', '.join(X_vars)}")
print(f"📈 TIPO DE REGRESIÓN: Regresión Lineal Múltiple (OLS)")


# =============================================================================
# 3. IMPLEMENTACIÓN Y ANÁLISIS DEL MODELO OLS
# =============================================================================
print("\n\n3. IMPLEMENTACIÓN DEL MODELO OLS")
print("-" * 40)

# 3.1. Agregar la constante (Intercepto o Beta_0)
X = sm.add_constant(X)

# 3.2. Ajustar el modelo de Regresión Lineal (OLS)
modelo_ols = sm.OLS(Y, X, hasconst=True).fit()

# 3.3. Imprimir el resumen de los resultados
print("\n**RESUMEN DEL MODELO OLS**")
print("===========================")
print(modelo_ols.summary())
```

# 4. 🧠 Interpretación Crítica del Modelo OLS (Tasa de Desempleo)

***
## 4.1. Validación de la Hipótesis del Proyecto

La hipótesis del proyecto es: **"La tasa de crecimiento del PIB influye mucho más que la inflación y el PIB actual de un país, en su tasa de desempleo."**

Al comparar la magnitud absoluta de los coeficientes, se determina que la hipótesis **se refuta** con los resultados de este modelo.

| Variable | Coeficiente ($\beta$) | P-valor (P>|t|) | Magnitud $|\beta|$ | Conclusión |
| :--- | :--- | :--- | :--- | :--- |
| **PIB (log)** | **+0.0600** | **0.042** | **0.0600** | **Mayor impacto y único significativo al 5%.** |
| **Crecimiento Económico** | -0.0165 | 0.287 | 0.0165 | Menor impacto. No es significativo. |
| **Inflación (CPI %)** | -0.0052 | 0.325 | 0.0052 | Impacto mínimo y no es significativo. |

**Conclusión Principal:** La variable con la mayor magnitud de impacto es el **PIB (log) ($\beta = 0.0600$)**, no el Crecimiento Económico. Solo esta variable es **estadísticamente significativa** ($p = 0.042$) al nivel de $\alpha=0.05$.

***
## 4.2. Crítica Estadística y Alineación Económica

### 📉 Evaluación del Ajuste (Resultado Negativo)

El modelo presenta un **ajuste extremadamente pobre**:

* **R-squared ($\mathbf{R^2}$):** **0.002** (0.2%). Este es un **resultado negativo** que indica que las variables macroeconómicas seleccionadas explican una porción insignificante de la variabilidad en la Tasa de Desempleo.
* **Significancia Global:** El **Prob (F-statistic)** es $\mathbf{0.0916}$, lo que implica que el modelo en su totalidad **no es estadísticamente significativo** al nivel de $\alpha=0.05$ (se acepta la hipótesis nula de que todos los $\beta$ son cero).

### 🔗 Conexión Teórica y Supuestos (TLC)

1.  **Coherencia Teórica:** El signo negativo del Crecimiento Económico ($\beta = -0.0165$) es **coherente** con la **Ley de Okun** (mayor crecimiento $\rightarrow$ menor desempleo), aunque su efecto es insignificante. El coeficiente positivo del **PIB (log) ($\beta = +0.0600$)** es, en cambio, **incoherente** con la intuición económica.
2.  **Teorema del Límite Central (TLC):**
    * A pesar de que el EDA previo mostró que las variables **rechazaban el supuesto de normalidad**, el **gran tamaño muestral** ($\mathbf{n=3,472}$) permite invocar el **Teorema del Límite Central (TLC)**.
    * Esto asegura que la distribución de los estimadores OLS es aproximadamente normal, lo que nos permite confiar en la robustez de los $P$-valores calculados. La principal conclusión es que la **relación lineal es estadísticamente débil**.
